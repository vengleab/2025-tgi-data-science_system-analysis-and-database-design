```python
# import matplotlib.pyplot as plt
from matplotlib import pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('./datasets/car-sales-extended.csv')
data.head(10)
```

```python
count_make = data["Make"].value_counts()
count_make
```

```python
count_make.index
```

```python
category = count_make.index
values = count_make.values

plt.pie(values, labels=category)
plt.show()

```

```python
count_make.index
```

```python
category = count_make.index
values = count_make.values

plt.pie(values, labels=category, colors=["white", "white", "white", "red"], autopct="%1.1f%%")
plt.show()
```

```python
category = count_make.index
values = count_make.values

plt.pie(values, labels=category, autopct="%1.1f%%")
plt.show()
```

# Chart Styling

## Hatch

```python
category = count_make.index
values = count_make.values

plt.pie(values, labels=category, autopct="%1.1f%%", hatch="o")
plt.show()
```

## Weight &  Height

```python
figure, ax = plt.subplots()
prices = data[["Price"]]
figure.set_figwidth(30)
figure.set_figheight(3)
plt.hist(prices)
plt.show()
```

```python
figure, ax = plt.subplots()
prices = data[["Price"]]
figure.set_figwidth(3)
figure.set_figheight(3)
ax.set_title("Histogram of Price")
ax.set_xlabel("Price")
ax.set_ylabel("Frequency / COUNT")
plt.hist(prices)
plt.show()
```

## Explode and Shadow

```python
category = count_make.index
values = count_make.values

plt.pie(values, labels=category,shadow=True)
plt.show()
```

```python
category
```

```python
category = count_make.index
values = count_make.values
explodes = [0, 0, 0, 0.3]

plt.pie(values, labels=category,shadow=True, explode=explodes)
plt.show()
```

```python
category = count_make.index
values = count_make.values
explodes = [0, 0, 0, 0.3]

plt.pie(values, labels=category,shadow=True, explode=explodes, startangle=90)
plt.show()
```

# Bar Chart

```python
data.head(10)
```

```python
sales = data[["Make", "Price"]].groupby("Make").sum()
sales
```

```python
sales.index
```

```python
fig, ax = plt.subplots()
plt.bar(sales.index, sales.Price)
```

```python
fig, ax = plt.subplots()
color=["red", "blue", "green", "yellow"]
idx = 0
for item in zip(sales.index, sales.Price):
    plt.bar(item[0], item[1], color=color[idx])
    idx += 1
# plt.bar(sales.index, sales.Price, color=["red", "blue", "green", "yellow"])
ax.legend()
plt.show()
```

```python
figure, ax = plt.subplots()
prices = data[["Price"]]
figure.set_figwidth(10)
figure.set_figheight(5)
ax.set_title("Histogram of Price")
ax.set_xlabel("Price")
ax.set_ylabel("Frequency / COUNT")
plt.hist(prices, bins=20)
plt.show()
```

```python
figure, ax = plt.subplots()
prices = data[["Price"]]
figure.set_figwidth(10)
figure.set_figheight(5)
ax.set_title("Histogram of Price")
ax.set_xlabel("Price")
ax.set_ylabel("Frequency / COUNT")
plt.hist(prices, bins=20)
ax.set_yticks([0, 600, 10000])
plt.show()
```
