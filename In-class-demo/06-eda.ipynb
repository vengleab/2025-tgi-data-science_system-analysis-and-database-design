# Is higher income to be approved

```python
df.columns
```

```python
df[["monthly_income", "loan_status"]]
```

```python
income_class = []

for i in df["monthly_income"]:
    if i > df["monthly_income"].mean():
        income_class.append("high")
    else:
        income_class.append("low")

df["monthly_income_class"] = income_class
df
```

```python
monthy_class_summary =df[["loan_status", "monthly_income_class"]].groupby("monthly_income_class").value_counts()
monthy_class_summary
```

```python
low_income = monthy_class_summary.loc["low"]
plt.bar([str(i) for i in low_income.index], low_income.values)
```

```python
high_income =monthy_class_summary.loc["high"]
high_income
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].pie( high_income.values, labels=[str(i) for i in high_income.index], autopct='%1.1f%%')
ax[0].title.set_text("High Income")
ax[1].pie(low_income.values, labels=[str(i) for i in low_income.index], autopct='%1.1f%%')
ax[1].title.set_text("Low Income")
plt.show()
```
